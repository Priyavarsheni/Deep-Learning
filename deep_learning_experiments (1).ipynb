```python
import numpy as np
import matplotlib.pyplot as plt

# Loss function
def loss(w):
    return (w - 3) ** 2

def gradient(w):
    return 2 * (w - 3)

lr = 0.1
epochs = 50
w = -5
losses = []

# Gradient Descent
for i in range(epochs):
    w = w - lr * gradient(w)
    losses.append(loss(w))

plt.plot(losses, label="Gradient Descent")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.title("Gradient Descent Optimization")
plt.legend()
plt.grid()
plt.show()

print("Final Weight:", w)
print("Final Loss:", loss(w))
```

```python
import numpy as np

# Input and target
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([[0], [1], [1], [0]])

# Initialize weights
np.random.seed(1)
W1 = np.random.randn(2, 4)
W2 = np.random.randn(4, 1)
lr = 0.1

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Training
for epoch in range(10000):
    # Forward propagation
    h = sigmoid(np.dot(X, W1))
    output = sigmoid(np.dot(h, W2))

    # Backpropagation
    error = y - output
    d_output = error * output * (1 - output)

    d_hidden = np.dot(d_output, W2.T) * h * (1 - h)

    W2 += lr * np.dot(h.T, d_output)
    W1 += lr * np.dot(X.T, d_hidden)

print("Predicted Output:")
print(np.round(output, 2))

import matplotlib.pyplot as plt

# Side-by-side bar chart to remove continuous line artifacts
x_indices = np.arange(len(y))
width = 0.35

plt.bar(x_indices - width/2, y.flatten(), width, label="Actual", color='royalblue')
plt.bar(x_indices + width/2, output.flatten(), width, label="Predicted", color='orange', alpha=0.8)

plt.title("Neural Network Output (XOR)")
plt.xlabel("Input Samples")
plt.ylabel("Value / Probability")
plt.xticks(x_indices, ["[0,0]", "[0,1]", "[1,0]", "[1,1]"])
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

Experiment 15: BPTT Flow and Gradient Analysis

```python
import tensorflow as tf
import matplotlib.pyplot as plt

x = tf.Variable([[0.2, 0.4, 0.6, 0.8]])
w = tf.Variable(0.5)

with tf.GradientTape() as tape:
    h = 0.0
    outputs = []
    for t in range(4):
        h = tf.tanh(h + x[0, t] * w)
        outputs.append(h)
    loss = (h - 1.0) ** 2

gradients = tape.gradient(loss, x)

print("Loss:", loss.numpy())
print("Gradients:", gradients.numpy())

plt.plot(range(1, 5), abs(gradients.numpy()[0]), 'o-')
plt.title("BPTT Gradient Analysis")
plt.xlabel("Time Step")
plt.ylabel("Gradient Magnitude")
plt.grid()
plt.show()
```

Experiment 17: RNN Gradient Analysis and Gradient Clipping

```python
import tensorflow as tf
import matplotlib.pyplot as plt

x = tf.random.normal((1, 5, 1))
model = tf.keras.Sequential([
    tf.keras.layers.SimpleRNN(8, input_shape=(5, 1)),
    tf.keras.layers.Dense(1)
])

with tf.GradientTape() as tape:
    y = model(x)
    loss = tf.reduce_mean(y ** 2)

grads = tape.gradient(loss, model.trainable_variables)

# Gradient clipping
clipped = [tf.clip_by_norm(g, 1.0) for g in grads]

print("Original gradient norm:",
      tf.linalg.global_norm(grads).numpy())
print("Clipped gradient norm:",
      tf.linalg.global_norm(clipped).numpy())

plt.bar(['Original', 'Clipped'],
        [tf.linalg.global_norm(grads).numpy(),
         tf.linalg.global_norm(clipped).numpy()])
plt.title("Gradient Clipping")
plt.ylabel("Gradient Norm")
plt.show()
```

Experiment 25: Restricted Boltzmann Machine (RBM) Energy and Reconstruction

```python
import numpy as np
import matplotlib.pyplot as plt

# Input data
v = np.array([[1, 0, 1, 0]])

# RBM weights and biases
W = np.random.rand(4, 2)
hb = np.zeros(2)
vb = np.zeros(4)

# Hidden layer
h_prob = 1 / (1 + np.exp(-(v @ W + hb)))
h = (h_prob > 0.5).astype(int)

# Reconstruct visible layer
v_prob = 1 / (1 + np.exp(-(h @ W.T + vb)))
reconstructed = (v_prob > 0.5).astype(int)

print("Original:", v)
print("Reconstructed:", reconstructed)

# Graph
plt.plot(v[0], 'o-', label='Original')
plt.plot(reconstructed[0], 's--', label='Reconstructed')
plt.title("RBM Reconstruction")
plt.legend()
plt.show()
```

Experiment 26: Gibbs Sampling and Contrastive Divergence

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

v = np.array([[1, 0, 1, 0]])
W = np.random.rand(4, 2) * 0.5
hb = np.zeros(2)
vb = np.zeros(4)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Gibbs sampling
h = (sigmoid(v @ W + hb) > 0.5).astype(int)
v_recon = (sigmoid(h @ W.T + vb) > 0.5).astype(int)

# Contrastive Divergence weight update
lr = 0.1
h_recon = (sigmoid(v_recon @ W + hb) > 0.5).astype(int)
W += lr * (v.T @ h - v_recon.T @ h_recon)

print("Original:", v)
print("Reconstructed:", v_recon)

plt.plot(v[0], 'o-', label='Original')
plt.plot(v_recon[0], 's--', label='Reconstructed')
plt.title("Gibbs Sampling and CD")
plt.legend()
plt.show()
```

Experiment 28: Deep Boltzmann Machine Simulation

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

# Visible and hidden layers
v = np.array([[1, 0, 1, 0]])
W1 = np.random.rand(4, 3)
W2 = np.random.rand(3, 2)

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Forward pass
h1 = (sigmoid(v @ W1) > 0.5).astype(int)
h2 = (sigmoid(h1 @ W2) > 0.5).astype(int)

# Reconstruction
h1_recon = (sigmoid(h2 @ W2.T) > 0.5).astype(int)
v_recon = (sigmoid(h1_recon @ W1.T) > 0.5).astype(int)

print("Original:", v)
print("Reconstructed:", v_recon)

plt.plot(v[0], 'o-', label='Original')
plt.plot(v_recon[0], 's--', label='Reconstructed')
plt.title("Boltzmann Machine Reconstruction")
plt.legend()
plt.show()
```

Experiment 29: VAE Sample Generation

```python
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras import layers, Model

# Encoder
inputs = layers.Input(shape=(28, 28, 1))
x = layers.Flatten()(inputs)
x = layers.Dense(64, activation='relu')(x)
z_mean = layers.Dense(2)(x)
z_log_var = layers.Dense(2)(x)

# Sampling
def sampling(args):
    mean, log_var = args
    eps = tf.random.normal(shape=tf.shape(mean))
    return mean + tf.exp(0.5 * log_var) * eps

z = layers.Lambda(sampling)([z_mean, z_log_var])
encoder = Model(inputs, z)

# Decoder
decoder_input = layers.Input(shape=(2,))
x = layers.Dense(64, activation='relu')(decoder_input)
x = layers.Dense(784, activation='sigmoid')(x)
outputs = layers.Reshape((28, 28, 1))(x)
decoder = Model(decoder_input, outputs)

# Generate sample
sample = decoder(tf.random.normal((1, 2)))

plt.imshow(sample[0, :, :, 0], cmap='gray')
plt.title("VAE Generated Sample")
plt.axis('off')
plt.show()
```

Experiment 30: Generative Model Output Comparison and Evaluation

```python
import tensorflow as tf
import matplotlib.pyplot as plt
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, Reshape, Input

# Simple generator (VAE)
vae = Sequential([
    Input(shape=(10,)),
    Dense(64, activation='relu'),
    Dense(784, activation='sigmoid'),
    Reshape((28, 28))
])

# Simple generator (GAN)
gan = Sequential([
    Input(shape=(10,)),
    Dense(128, activation='relu'),
    Dense(784, activation='sigmoid'),
    Reshape((28, 28))
])

# Generate images
noise = tf.random.normal((1, 10))
vae_img = vae(noise)[0]
gan_img = gan(noise)[0]

# Display comparison
plt.subplot(1, 2, 1)
plt.imshow(vae_img, cmap='gray')
plt.title("VAE Output")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(gan_img, cmap='gray')
plt.title("GAN Output")
plt.axis('off')

plt.show()
```
